```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, Input, Lambda, GlobalAveragePooling2D, Concatenate
from tensorflow.keras.applications import ResNet50, MobileNetV3Small
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, classification_report

# Convolutional Block Attention Module (CBAM) Layer
class CBAM(tf.keras.layers.Layer):
    def get_config(self):
        config = super().get_config()
        config.update({
            "filters": self.filters,"reduction_ratio":
            self.reduction_ratio,
        })
        return config
        
    def __init__(self, filters, reduction_ratio=8):
        super(CBAM, self).__init__()
        self.filters = filters
        self.reduction_ratio = reduction_ratio

    def build(self, input_shape):
        self.avg_pool = tf.keras.layers.GlobalAveragePooling2D()
        self.max_pool = tf.keras.layers.GlobalMaxPooling2D()
        self.shared_dense_one = tf.keras.layers.Dense(self.filters // self.reduction_ratio, activation='relu')
        self.shared_dense_two = tf.keras.layers.Dense(self.filters)
        self.conv_spatial = tf.keras.layers.Conv2D(1, kernel_size=7, padding='same', activation='sigmoid')

    def call(self, input_feature):
        avg_pool = self.avg_pool(input_feature)
        max_pool = self.max_pool(input_feature)
        avg_out = self.shared_dense_two(self.shared_dense_one(avg_pool))
        max_out = self.shared_dense_two(self.shared_dense_one(max_pool))
        channel_attention = tf.nn.sigmoid(avg_out + max_out)
        channel_attention = tf.expand_dims(tf.expand_dims(channel_attention, 1), 1)
        channel_refined = input_feature * channel_attention

        avg_pool_spatial = tf.reduce_mean(channel_refined, axis=-1, keepdims=True)
        max_pool_spatial = tf.reduce_max(channel_refined, axis=-1, keepdims=True)
        spatial_attention = self.conv_spatial(tf.concat([avg_pool_spatial, max_pool_spatial], axis=-1))

        return channel_refined * spatial_attention


# Path ke folder dataset
train_dataset_path = r'C:\Users\Asus\PENELITIAN_RINI_VAMULIA_OA_CNN\Processed_TrainOA\Final'
val_dataset_path = r'C:\Users\Asus\PENELITIAN_RINI_VAMULIA_OA_CNN\Processed_ValOA\Final'

# Data Augmentation untuk training
train_datagen = ImageDataGenerator(
    rescale=1.0 / 127.5 - 1.0,  # Normalisasi ke [-1,1]
    width_shift_range=0.1,
    height_shift_range=0.1,
    rotation_range=15,
    zoom_range=0.18,
    brightness_range=[0.8, 1.2],  # Variasi kecerahan
    horizontal_flip=True,
    fill_mode='nearest'  # Mengisi area yang hilang
)

# Data preprocessing untuk validasi (tanpa augmentasi)    
val_datagen = ImageDataGenerator(rescale=1.0 / 127.5 - 1.0)

# Load train dataset
train_generator = train_datagen.flow_from_directory(
    train_dataset_path,
    target_size=(224, 224),
    color_mode='rgb',
    batch_size=32,
    class_mode='categorical',
    shuffle=True
)

# Load validation dataset
val_generator = val_datagen.flow_from_directory(
    val_dataset_path,
    target_size=(224, 224),
    color_mode='rgb',
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

# Cek jumlah gambar per kelas
def count_images_per_class(directory):
    for class_name in os.listdir(directory):
        class_path = os.path.join(directory, class_name)
        if os.path.isdir(class_path):
            num_images = len([
                name for name in os.listdir(class_path)
                if os.path.isfile(os.path.join(class_path, name))
            ])
            print(f"{class_name}: {num_images}")

print("Jumlah Gambar Setiap Grade Training:")
count_images_per_class(train_dataset_path)
print("\nJumlah Gambar Setiap Grade Validasi:")
count_images_per_class(val_dataset_path)
    
# Load Model Architecture MobileNetV3 dan ResNet50
input_tensor = Input(shape=(224, 224, 3))

mobilenet = MobileNetV3Small(weights='imagenet', include_top=False, input_tensor=input_tensor)
resnet = ResNet50(weights='imagenet', include_top=False, input_tensor=input_tensor)

# Freeze lebih banyak layer awal di ResNet
for layer in resnet.layers[:80]:
    layer.trainable = False

for layer in mobilenet.layers[:50]:
    layer.trainable = False

# Menambahkan Multi-CBAM
cbam_mobilenet = CBAM(filters=576)(mobilenet.output)
cbam_mobilenet = CBAM(filters=576)(cbam_mobilenet)

cbam_resnet = CBAM(filters=2048)(resnet.output)
cbam_resnet = CBAM(filters=2048)(cbam_resnet)

# Global Average Pooling
x1 = GlobalAveragePooling2D()(cbam_mobilenet)
x2 = GlobalAveragePooling2D()(cbam_resnet)
merged = Concatenate()([x1, x2])

x = Dense(512, activation='relu')(merged)
x = Dropout(0.7)(x)
output = Dense(5, activation='softmax')(x)

model = Model(inputs=input_tensor, outputs=output)

# Compile Model untuk Transfer learning
tf.keras.utils.set_random_seed(42)  # Reproducibility
model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('best_multicbam.h5', save_best_only=True, monitor='val_loss', mode='min'),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7)
]

# Tampilkan arsitektur model
model.summary()

# Training Model untuk Transfer learning
history_transfer_learning = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10,
    callbacks=callbacks
)

# Fine-Tuning MobileNetV3Small
# Freeze semua layer di MobileNetV3Small terlebih dahulu
for layer in mobilenet.layers:
    layer.trainable = False

# Unfreeze Seluruh Layer MobileNetV3Small
for layer in mobilenet.layers:
    layer.trainable = True

# Compile model untuk Fine-Tuning
tf.keras.utils.set_random_seed(42)  # Reproducibility
model.compile(
    optimizer=Adam(learning_rate=1e-7),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Training Fine-Tuning
history_fine_tuning = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=15,
    callbacks=callbacks
)

# Plot hasil training
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history_transfer_learning.history['accuracy'], label='Train Accuracy')
plt.plot(history_transfer_learning.history['val_accuracy'], label='Validation Accuracy')
plt.plot(history_fine_tuning.history['accuracy'], label='Train Accuracy')
plt.plot(history_fine_tuning.history['val_accuracy'], label='Validation Accuracy')
plt.legend()
plt.title('Accuracy')

plt.subplot(1, 2, 2)
plt.plot(history_transfer_learning.history['loss'], label='Train Loss')
plt.plot(history_transfer_learning.history['val_loss'], label='Validation Loss')
plt.plot(history_fine_tuning.history['loss'], label='Train Loss')
plt.plot(history_fine_tuning.history['val_loss'], label='Validation Loss')
plt.legend()
plt.title('Loss')
plt.show()


# Evaluasi Model
y_pred = model.predict(val_generator)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = val_generator.classes
class_labels = list(val_generator.class_indices.keys())

# Evaluasi dengan confusion matrix
y_true = val_generator.classes
y_pred = np.argmax(model.predict(val_generator), axis=1)
class_labels = list(val_generator.class_indices.keys())
conf_matrix = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

# Classification Report
print(classification_report(y_true, y_pred, target_names=class_labels))
```

```python
from tensorflow.keras.models import load_model
import tensorflow as tf
import numpy as np
import cv2
import os
from keras.models import load_model
from tensorflow.keras.losses import Loss

class FocalLoss(Loss):
    def __init__(self, gamma=2.0, alpha=0.25, **kwargs):
        super(FocalLoss, self).__init__(**kwargs)
        self.gamma = gamma
        self.alpha = alpha

    def call(self, y_true, y_pred):
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1. - 1e-7)
        cross_entropy = -y_true * tf.math.log(y_pred)
        weight = self.alpha * tf.math.pow(1 - y_pred, self.gamma)
        loss = weight * cross_entropy
        return tf.reduce_sum(loss, axis=1)

# Definisi ulang Layer CBAM
class CBAM(tf.keras.layers.Layer):
    def get_config(self):
        config = super().get_config()
        config.update({
            "filters": self.filters,
            "reduction_ratio": self.reduction_ratio,
        })
        return config

    def __init__(self, filters, reduction_ratio=8, **kwargs):
        super(CBAM, self).__init__(**kwargs)
        self.filters = filters
        self.reduction_ratio = reduction_ratio

    def build(self, input_shape):
        self.avg_pool = tf.keras.layers.GlobalAveragePooling2D()
        self.max_pool = tf.keras.layers.GlobalMaxPooling2D()
        self.shared_dense_one = tf.keras.layers.Dense(self.filters // self.reduction_ratio, activation='relu')
        self.shared_dense_two = tf.keras.layers.Dense(self.filters)
        self.conv_spatial = tf.keras.layers.Conv2D(1, kernel_size=7, padding='same', activation='sigmoid')

    def call(self, input_feature):
        avg_pool = self.avg_pool(input_feature)
        max_pool = self.max_pool(input_feature)
        avg_out = self.shared_dense_two(self.shared_dense_one(avg_pool))
        max_out = self.shared_dense_two(self.shared_dense_one(max_pool))
        channel_attention = tf.nn.sigmoid(avg_out + max_out)
        channel_attention = tf.expand_dims(tf.expand_dims(channel_attention, 1), 1)
        channel_refined = input_feature * channel_attention

        avg_pool_spatial = tf.reduce_mean(channel_refined, axis=-1, keepdims=True)
        max_pool_spatial = tf.reduce_max(channel_refined, axis=-1, keepdims=True)
        spatial_attention = self.conv_spatial(tf.concat([avg_pool_spatial, max_pool_spatial], axis=-1))

        return channel_refined * spatial_attention

# Load model dengan custom layer CBAM
model = load_model(
    "C:/Users/Asus/PENELITIAN_RINI_VAMULIA_OA_CNN/Architectural_experiments/morphology/best_multicbam.h5",
    custom_objects={'CBAM': CBAM, 'FocalLoss': FocalLoss}
)

# Fungsi preprocessing gambar (grayscale atau RGB akan dikonversi ke RGB 3-channel)
def preprocess_image(image_path, target_size=(224, 224)):
    img = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

    # Deteksi apakah grayscale → konversi ke RGB
    if len(img.shape) == 2 or img.shape[2] == 1:
        img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
    else:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    img = cv2.resize(img, target_size)
    img = img.astype('float32') / 255.0
    img = np.expand_dims(img, axis=0)  # Shape: (1, 224, 224, 3)

    return img

# Path gambar yang ingin diprediksi
image_path = "C:/Users/Asus/PENELITIAN_RINI_VAMULIA_OA_CNN/Processed_TestOA/Final/Grade0/final_9172581L.png"

# Preprocessing & Prediksi
img = preprocess_image(image_path)
pred = model.predict(img)

# Output hasil
print("Predicted probabilities:", pred)
print("Predicted class (GRADE):", np.argmax(pred))
```

```python
import numpy as np
from tensorflow.keras.preprocessing import image
from tensorflow.keras.models import load_model
import tensorflow as tf

# Fungsi preprocessing (pastikan grayscale & normalisasi ke [-1, 1])
def preprocess_image(image_path):
    img = image.load_img(image_path, target_size=(224, 224), color_mode='grayscale')
    img_array = image.img_to_array(img)
    img_array = img_array / 127.5 - 1  # normalisasi dari [0, 255] ke [-1, 1]
    img_array = np.expand_dims(img_array, axis=0)  # tambah batch dimensi
    return img_array

# Load model
model = load_model("C:/Users/Asus/PENELITIAN_RINI_VAMULIA_OA_CNN/Architectural_experiments/best_vgg16_model.h5")

# Path gambar yang ingin diprediksi
image_path = "C:/Users/Asus/PENELITIAN_RINI_VAMULIA_OA_CNN/Processed_TestOA/Morphological/Grade3/morph_9055836L.png"

# Preprocessing dan prediksi
img = preprocess_image(image_path)
pred = model.predict(img)

# Output hasil prediksi
class_names = ['GRADE 0', 'GRADE 1', 'GRADE 2', 'GRADE 3', 'GRADE 4']
predicted_class = np.argmax(pred)
predicted_proba = pred[0]

print("Probabilitas untuk tiap kelas:")
for i, prob in enumerate(predicted_proba):
    print(f"{class_names[i]}: {prob:.4f}")

print("\nPrediksi akhir:", class_names[predicted_class])
```
